```python
from ultralytics import YOLO
# 1. Cargar un modelo pre-entrenado (YOLOv8 nano es rápido y ligero)
model = YOLO('yolov8n.pt')
# 2. Entrenar el modelo
results = model.train(
data='dataset-piedra-papel-tijeras/data.yaml',
epochs=50, # Número de vueltas al dataset
imgsz=640, # Tamaño de la imagen
plots=True
)
```

```python
"""
Visualización AUTOMÁTICA de Resultados - Piedra Papel Tijeras
Este script encuentra automáticamente el último experimento
"""

import os
from IPython.display import Image, display
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================================
# AUTO-DETECTAR ÚLTIMO EXPERIMENTO
# ============================================================================

def encontrar_ultimo_experimento():
    """Encuentra automáticamente el último experimento de entrenamiento"""
    
    runs_path = Path('runs/detect')
    
    if not runs_path.exists():
        print(f"❌ No existe la carpeta {runs_path}")
        return None
    
    # Listar todos los experimentos
    experimentos = [d for d in runs_path.iterdir() if d.is_dir()]
    
    if not experimentos:
        print("❌ No se encontraron experimentos")
        return None
    
    # Ordenar por fecha de modificación (el más reciente primero)
    experimentos.sort(key=lambda x: x.stat().st_mtime, reverse=True)
    
    print("📂 Experimentos disponibles (más reciente primero):")
    print("=" * 70)
    for i, exp in enumerate(experimentos[:10], 1):  # Mostrar los 10 más recientes
        mtime = exp.stat().st_mtime
        from datetime import datetime
        fecha = datetime.fromtimestamp(mtime).strftime('%Y-%m-%d %H:%M:%S')
        print(f"   {i}. {exp.name:<20} - {fecha}")
    
    return experimentos[0]

# Encontrar el último experimento
ultimo_exp = encontrar_ultimo_experimento()

if ultimo_exp:
    results_path = str(ultimo_exp)
    print(f"\n✅ Usando experimento: {results_path}")
else:
    print("❌ No se pudo encontrar ningún experimento")
    exit()

# ============================================================================
# VISUALIZACIÓN DE RESULTADOS
# ============================================================================

print("\n" + "=" * 70)
print("🎯 ANÁLISIS DE RESULTADOS - PIEDRA PAPEL TIJERAS")
print("=" * 70)
print(f"📁 Ruta: {results_path}\n")

# ============================================================================
# 1. VISUALIZACIONES PRINCIPALES
# ============================================================================

print("📊 VISUALIZACIONES DEL ENTRENAMIENTO")
print("=" * 70)

imgs = {
    '📈 Curvas de entrenamiento': 'results.png',
    '🎯 Predicciones en validación': 'val_batch0_pred.jpg',
    '✅ Ground Truth (etiquetas)': 'val_batch0_labels.jpg',
    '🔢 Matriz de confusión': 'confusion_matrix.png',
    '📊 Matriz normalizada': 'confusion_matrix_normalized.png',
    '📉 Curva F1-Confidence': 'F1_curve.png',
    '📉 Curva Precision': 'P_curve.png',
    '📉 Curva Recall': 'R_curve.png',
    '📉 Curva PR': 'PR_curve.png',
}

imagenes_encontradas = 0
for titulo, archivo in imgs.items():
    path = f'{results_path}/{archivo}'
    if os.path.exists(path):
        print(f"\n{titulo}")
        print("-" * 70)
        try:
            display(Image(filename=path, width=800))
            imagenes_encontradas += 1
        except Exception as e:
            print(f"   ⚠️  Error mostrando imagen: {e}")
            print(f"   Disponible en: {path}")

if imagenes_encontradas == 0:
    print("\n⚠️  No se encontraron imágenes de visualización")

# ============================================================================
# 2. MÉTRICAS FINALES
# ============================================================================

print("\n" + "=" * 70)
print("📈 MÉTRICAS FINALES DEL MODELO")
print("=" * 70)

csv_path = f'{results_path}/results.csv'
if os.path.exists(csv_path):
    try:
        # Leer resultados
        df = pd.read_csv(csv_path)
        
        # Limpiar nombres de columnas (eliminar espacios)
        df.columns = df.columns.str.strip()
        
        # Mostrar columnas disponibles (para debug)
        print("\n📋 Columnas disponibles:")
        print(", ".join(df.columns.tolist()))
        
        # Última época
        ultima_epoca = df.iloc[-1]
        
        print(f"\n✨ Resumen del Entrenamiento:")
        print(f"   Total de épocas: {int(ultima_epoca['epoch'])}")
        
        # Métricas de detección
        print(f"\n🎯 Métricas de Detección:")
        
        # Buscar columnas de métricas (pueden tener espacios o no)
        map50_cols = [c for c in df.columns if 'mAP50(B)' in c or 'mAP50-95(B)' in c]
        
        if 'metrics/mAP50(B)' in df.columns:
            print(f"   mAP@50:     {ultima_epoca['metrics/mAP50(B)']:.4f}  ← Precisión principal")
            print(f"   mAP@50-95:  {ultima_epoca['metrics/mAP50-95(B)']:.4f}  ← Precisión promedio")
        elif len(map50_cols) > 0:
            for col in map50_cols:
                print(f"   {col}: {ultima_epoca[col]:.4f}")
        
        # Precision y Recall
        if 'metrics/precision(B)' in df.columns:
            print(f"   Precision:  {ultima_epoca['metrics/precision(B)']:.4f}")
        if 'metrics/recall(B)' in df.columns:
            print(f"   Recall:     {ultima_epoca['metrics/recall(B)']:.4f}")
        
        # Pérdidas
        print(f"\n📉 Pérdidas Finales:")
        loss_cols = [c for c in df.columns if 'loss' in c.lower()]
        for col in loss_cols:
            if 'train' in col:
                print(f"   {col}: {ultima_epoca[col]:.4f}")
        
        # ====================================================================
        # 3. GRÁFICOS PERSONALIZADOS
        # ====================================================================
        
        print("\n" + "=" * 70)
        print("📊 GRÁFICOS DE EVOLUCIÓN")
        print("=" * 70)
        
        fig, axes = plt.subplots(2, 2, figsize=(16, 10))
        fig.suptitle('Evolución del Entrenamiento - Piedra Papel Tijeras', 
                     fontsize=16, fontweight='bold')
        
        # Gráfico 1: mAP
        ax = axes[0, 0]
        if 'metrics/mAP50(B)' in df.columns:
            ax.plot(df['epoch'], df['metrics/mAP50(B)'], 
                   label='mAP@50', marker='o', linewidth=2, markersize=4)
            ax.plot(df['epoch'], df['metrics/mAP50-95(B)'], 
                   label='mAP@50-95', marker='s', linewidth=2, markersize=4)
            ax.set_xlabel('Época', fontsize=11)
            ax.set_ylabel('mAP', fontsize=11)
            ax.set_title('📈 Precisión Media (mAP)', fontsize=12, fontweight='bold')
            ax.legend(fontsize=10)
            ax.grid(True, alpha=0.3)
        
        # Gráfico 2: Precision & Recall
        ax = axes[0, 1]
        if 'metrics/precision(B)' in df.columns and 'metrics/recall(B)' in df.columns:
            ax.plot(df['epoch'], df['metrics/precision(B)'], 
                   label='Precision', marker='o', color='blue', linewidth=2, markersize=4)
            ax.plot(df['epoch'], df['metrics/recall(B)'], 
                   label='Recall', marker='s', color='orange', linewidth=2, markersize=4)
            ax.set_xlabel('Época', fontsize=11)
            ax.set_ylabel('Valor', fontsize=11)
            ax.set_title('🎯 Precision y Recall', fontsize=12, fontweight='bold')
            ax.legend(fontsize=10)
            ax.grid(True, alpha=0.3)
        
        # Gráfico 3: Losses
        ax = axes[1, 0]
        if 'train/box_loss' in df.columns:
            ax.plot(df['epoch'], df['train/box_loss'], 
                   label='Box Loss', marker='o', linewidth=2, markersize=4)
            if 'train/cls_loss' in df.columns:
                ax.plot(df['epoch'], df['train/cls_loss'], 
                       label='Class Loss', marker='s', linewidth=2, markersize=4)
            if 'train/dfl_loss' in df.columns:
                ax.plot(df['epoch'], df['train/dfl_loss'], 
                       label='DFL Loss', marker='^', linewidth=2, markersize=4)
            ax.set_xlabel('Época', fontsize=11)
            ax.set_ylabel('Loss', fontsize=11)
            ax.set_title('📉 Pérdidas de Entrenamiento', fontsize=12, fontweight='bold')
            ax.legend(fontsize=10)
            ax.grid(True, alpha=0.3)
        
        # Gráfico 4: Learning Rate
        ax = axes[1, 1]
        lr_cols = [c for c in df.columns if 'lr/' in c]
        if lr_cols:
            ax.plot(df['epoch'], df[lr_cols[0]], 
                   marker='o', color='green', linewidth=2, markersize=4)
            ax.set_xlabel('Época', fontsize=11)
            ax.set_ylabel('Learning Rate', fontsize=11)
            ax.set_title('📊 Tasa de Aprendizaje', fontsize=12, fontweight='bold')
            ax.grid(True, alpha=0.3)
        
        plt.tight_layout()
        
        # Guardar gráfico
        output_plot = f'{results_path}/evolucion_metricas.png'
        plt.savefig(output_plot, dpi=150, bbox_inches='tight')
        print(f"✅ Gráfico guardado: {output_plot}")
        
        plt.show()
        
        # ====================================================================
        # 4. TABLA DE PROGRESO
        # ====================================================================
        
        print("\n" + "=" * 70)
        print("📊 TABLA DE PROGRESO (Cada 10 épocas)")
        print("=" * 70)
        
        # Mostrar cada 10 épocas
        step = max(1, len(df) // 10)
        df_sample = df.iloc[::step]
        
        print(f"\n{'Época':<8} {'mAP50':<10} {'mAP50-95':<12} {'Box Loss':<12} {'Cls Loss':<12}")
        print("-" * 70)
        
        for _, row in df_sample.iterrows():
            epoca = int(row['epoch'])
            map50 = row.get('metrics/mAP50(B)', 0)
            map5095 = row.get('metrics/mAP50-95(B)', 0)
            box_loss = row.get('train/box_loss', 0)
            cls_loss = row.get('train/cls_loss', 0)
            
            print(f"{epoca:<8} {map50:<10.4f} {map5095:<12.4f} {box_loss:<12.4f} {cls_loss:<12.4f}")
        
    except Exception as e:
        print(f"⚠️  Error al procesar métricas: {e}")
        import traceback
        traceback.print_exc()
else:
    print(f"⚠️  No se encontró el archivo de resultados: {csv_path}")

# ============================================================================
# 5. INFORMACIÓN DEL MODELO
# ============================================================================

print("\n" + "=" * 70)
print("💾 ARCHIVOS DEL MODELO")
print("=" * 70)

weights_path = Path(results_path) / 'weights'
if weights_path.exists():
    best_model = weights_path / 'best.pt'
    last_model = weights_path / 'last.pt'
    
    if best_model.exists():
        size_mb = best_model.stat().st_size / (1024 * 1024)
        print(f"\n✅ Mejor modelo: {best_model}")
        print(f"   Tamaño: {size_mb:.2f} MB")
        print(f"\n💡 Para usar el modelo:")
        print(f"   from ultralytics import YOLO")
        print(f"   model = YOLO('{best_model}')")
        print(f"   results = model.predict('imagen.jpg', conf=0.25)")
        print(f"\n   # Ver resultados:")
        print(f"   for r in results:")
        print(f"       for box in r.boxes:")
        print(f"           clase = model.names[int(box.cls[0])]")
        print(f"           confianza = float(box.conf[0])")
        print(f"           print(f'{{clase}}: {{confianza:.2%}}')")
    
    if last_model.exists():
        size_mb = last_model.stat().st_size / (1024 * 1024)
        print(f"\n📦 Último checkpoint: {last_model}")
        print(f"   Tamaño: {size_mb:.2f} MB")
else:
    print("\n⚠️  No se encontró la carpeta de pesos")

# ============================================================================
# RESUMEN FINAL
# ============================================================================

print("\n" + "=" * 70)
print("🎉 ANÁLISIS COMPLETADO")
print("=" * 70)
print(f"\n📁 Resultados en: {results_path}")
print(f"\n📂 Archivos generados:")
print(f"   ├── weights/best.pt              ← Mejor modelo")
print(f"   ├── weights/last.pt              ← Último checkpoint")
print(f"   ├── results.csv                  ← Métricas por época")
print(f"   ├── results.png                  ← Gráficos oficiales")
print(f"   ├── evolucion_metricas.png       ← Gráfico personalizado")
print(f"   ├── confusion_matrix.png         ← Matriz de confusión")
print(f"   └── val_batch0_pred.jpg          ← Predicciones visuales")

print("\n💡 Clases detectadas:")
print("   0: Piedra ✊")
print("   1: Papel 🖐")
print("   2: Tijeras ✌️")

print("\n✨ ¡Todo listo para hacer predicciones!")
```
