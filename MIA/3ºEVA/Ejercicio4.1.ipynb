Preparación inicial del texto

```python
# Definimos nuestra variable de texto
texto = "Tengo un perro verde y un gato azul"
# Convertimos a minúsculas
texto_minusculas = texto.lower()
print(f"Texto en minúsculas: {texto_minusculas}")
# División básica con split (basada en espacios)
tokens_basicos = texto_minusculas.split()
print(f"Tokens con split(): {tokens_basicos}")
print(f"Número de tokens: {len(tokens_basicos)}")
```

Configuración de NLTK

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
# Descargamos los recursos necesarios
nltk.download('punkt') # punkt es el "cerebro" que le enseña a NLTK a
#dividir un texto en palabras y frases de forma gramaticalmente
#correcta, y no solo cortando por espacios.
nltk.download('stopwords') #palabras irrelevantes
```

Opciones disponibles en NLTK

```python
# Listar todos los idiomas/opciones disponibles en el corpus de
stopwords
print("Idiomas y opciones disponibles en NLTK Stopwords:")
print(stopwords.fileids())
```

Configuración en Español

```python
# Definimos el conjunto de palabras vacías en español
stop_words = set(stopwords.words('spanish'))
# Imprimimos la lista para que los alumnos la vean
print("Lista de Stop Words en español:")
print(stop_words)
```

Instalación necesaria para hacer la tokenización avanzada

```python
import nltk
nltk.download('punkt_tab')
```

Tokenización avanzada

```python
# Tokenizamos el texto original (en minúsculas)
tokens = word_tokenize(texto_minusculas)
print("Resultado de word_tokenize:")
print(tokens)
```

Filtrado de Stop Words

```python
# Filtramos: "Crea una lista con cada palabra de 'tokens' si esa
#palabra NO está en 'stop_words'"
texto_filtrado = [word for word in tokens if word not in stop_words]
print("Resultado final (sin Stop Words):")
print(texto_filtrado)
# Comparativa de longitud
print(f"\nLongitud original: {len(tokens)} palabras")
print(f"Longitud tras filtrado: {len(texto_filtrado)} palabras")
```

Stemming (Lematización de raíces o Truncamiento)

```python
#Import nltk
# wordnet es un diccionario inteligente que le permite a NLTK saber
#que un 'perrito' es un 'animal' y que 'corriendo' es una acción
#derivada de 'correr'."
nltk.download('wordnet')
from nltk.stem import SnowballStemmer
# Configuramos el stemmer para español
stemmer = SnowballStemmer('spanish')
# Lista de palabras de prueba (verbos y sustantivos)
palabras = ['correr', 'corriendo', 'corredor', 'perros', 'perrito',
'casas', 'casita']
print("--- RESULTADO DEL STEMMING ---")
for p in palabras:
    raiz = stemmer.stem(p)
    print(f"{p} -> {raiz}")
```

Lemmatizar en Español (usando spaCy)

```python
import spacy
# Cargamos el modelo en español
nlp = spacy.load('es_core_news_sm')
# Nuestra frase original
texto = "Tengo un perro verde y un gato azul. Los perros están corriendo."
# Procesamos el texto
doc = nlp(texto)
print("--- RESULTADO DE LA LEMMATIZACIÓN ---")
# Imprimimos: Palabra original -> Lema (Raíz de diccionario)
for token in doc:
    print(f"{token.text} -> {token.lemma_}")
```

Comparar ambos procesos en un frase

```python
frase = "Los gatos están durmiendo en las alfombras"
tokens = word_tokenize(frase.lower()) # Usando lo que aprendimos antes
print(f"{'PALABRA':<15} | {'STEMMING':<15} | {'LEMA'}")
print("-" * 45)
for t in tokens:
# Stemming con NLTK
    st = stemmer.stem(t)
# Lema con spaCy (procesamos la palabra individualmente)
lm = nlp(t)[0].lemma_
print(f"{t:<15} | {st:<15} | {lm}")
```

Encontrar Sinónimos y Antónimos (Synsets)

```python
from nltk.corpus import wordnet
# Buscamos el concepto de "perro" (dog)
syns = wordnet.synsets("dog")
# Ver el nombre del concepto y su definición
print(f"Concepto: {syns[0].name()}")
print(f"Definición: {syns[0].definition()}")
# Ver sinónimos
print(f"Sinónimos: {syns[0].lemma_names()}")
```
