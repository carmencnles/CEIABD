```python
import os
from dotenv import load_dotenv
# Cargar variables del archivo .env
load_dotenv()
# La variable queda disponible en el sistema automáticamente
```

```python
import librosa
import soundfile as sf

# 1. Cargar el audio y convertirlo a 16KHz y Mono automáticamente
# 'sr=16000' fija la frecuencia, 'mono=True' combina los canales 
y, sr = librosa.load("podcast.mp3", sr=16000, mono=True)

# 2. Guardar el archivo optimizado
sf.write("podcast_optimizado.mp3", y, sr)

print("¡Audio optimizado a 16KHz y Mono con éxito!")
```

```python
import os
from groq import Groq
# Inicializar el cliente de Groq
client = Groq()
# Definir la ruta del archivo de audio
filename = "podcast_optimizado.mp3"
# Proceso de apertura y transcripción
with open(filename, "rb") as file:
    transcription = client.audio.transcriptions.create(
        file=file, # Archivo de audio binario
        model="whisper-large-v3-turbo", # Modelo seleccionado
        prompt="Contexto o palabras clave", # Opcional para mejorar precisión
        response_format="verbose_json", # Formato detallado de salida
        language="es", # Idioma del audio (ISO-639-1)
    temperature=0.0 # Mayor determinismo (0.0)
)
# Resultado final
print(transcription.text)
```

```python
# Usando otro modelo de Groq para traducir el texto obtenido
completion = client.chat.completions.create(
    model="llama-3.3-70b-versatile", # Modelo versátil mencionado en el PDF 
    messages=[
        {"role": "system", "content": "Eres un traductor profesional."},
        {"role": "user", "content": f"Traduce este texto al inglés: {transcription.text[:500]}..."}
    ]
)
print("\n--- Traducción (Primeros 500 caracteres) ---")
print(completion.choices[0].message.content)
```

```python
import nltk
from nltk.tokenize import word_tokenize

# 1. Aseguramos que los recursos de NLTK estén descargados
nltk.download('punkt_tab')

# 2. El texto que obtuvimos de Groq
texto_podcast = transcription.text

# 3. Limpieza básica: pasar a minúsculas
texto_minusculas = texto_podcast.lower()

# 4. Tokenización (separar en palabras)
tokens = word_tokenize(texto_minusculas)

# 5. Mostrar resultados
print("\n--- Análisis de Tokenización ---")
print(f"Número total de palabras (tokens): {len(tokens)}")
print("Primeros 10 tokens:")
print(tokens[:10])
```

```python
import nltk
from nltk.corpus import stopwords
from collections import Counter

# 1. Descargar la lista de palabras vacías (Stopwords)
nltk.download('stopwords')

# 2. Definir el idioma (Español)
palabras_vacias = set(stopwords.words('spanish'))

# 3. Filtrar los tokens: quitamos palabras vacías y signos de puntuación
tokens_limpios = [word for word in tokens if word.isalnum() and word not in palabras_vacias]

# 4. Contar las palabras más frecuentes
frecuencia = Counter(tokens_limpios)

print("\n--- Análisis de Contenido ---")
print(f"Palabras tras la limpieza: {len(tokens_limpios)}")
print("Las 5 palabras más importantes del podcast:")
for palabra, conteo in frecuencia.most_common(5):
    print(f"- {palabra}: {conteo} veces")
```
