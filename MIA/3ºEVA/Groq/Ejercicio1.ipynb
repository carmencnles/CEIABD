```python
import os
from groq import Groq
# Inicializar el cliente de Groq
client = Groq()
# Definir la ruta del archivo de audio
filename = "podcast.mp3"
# Proceso de apertura y transcripción
with open(filename, "rb") as file:
    transcription = client.audio.transcriptions.create(
        file=file, # Archivo de audio binario
        model="whisper-large-v3-turbo", # Modelo seleccionado
        prompt="Contexto o palabras clave", # Opcional para mejorar precisión
        response_format="verbose_json", # Formato detallado de salida
        language="es", # Idioma del audio (ISO-639-1)
    temperature=0.0 # Mayor determinismo (0.0)
)
# Resultado final
print(transcription.text)
```
