```python
# 1. Instalación de la librería de Texto a Voz
!pip install gTTS

from gtts import gTTS
import os

# Configuramos los archivos
archivo_texto = "transcripcion_resultado.txt"
archivo_audio_nuevo = "voz_final.mp3"

# 2. Leer el texto generado por Whisper
if os.path.exists(archivo_texto):
    with open(archivo_texto, "r", encoding="utf-8") as f:
        contenido = f.read()
    
    if contenido.strip():
        print("Generando audio a partir del texto...")
        
        # 3. Crear el objeto de voz en español
        # Puedes cambiar 'es' por 'en' si quieres que suene con acento inglés
        tts = gTTS(text=contenido, lang='es')
        
        # 4. Guardar el nuevo archivo de audio
        tts.save(archivo_audio_nuevo)
        print(f"¡Listo! El audio se ha guardado como: {archivo_audio_nuevo}")
    else:
        print("El archivo de texto está vacío.")
else:
    print(f"No se encontró el archivo {archivo_texto}. Primero debes completar la parte de Whisper.")
```
