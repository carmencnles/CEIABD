```python
import os
import time  
from groq import Groq
from google import genai
from dotenv import load_dotenv

# Carga las claves del archivo .env 
load_dotenv()

# Inicialización de clientes 
client_groq = Groq()
client_gemini = genai.Client()

tema = input("¿Qué quieres comparar? (Cualquier tema): ")

# --- MEDICIÓN DE GROQ (Ejercicio 1: Impacto del Rol System) --- 
start_groq = time.time() 
try:
    res_groq = client_groq.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system", 
                "content": "Eres un programador senior amargado que odia responder preguntas básicas. Tu tono es pasivo-agresivo, usas ironía y eres extremadamente breve."
            }, # El rol system define la personalidad 
            {"role": "user", "content": tema} # El user dispara la acción 
        ]
    )
    end_groq = time.time() 
    tiempo_groq = end_groq - start_groq

    print(f"\n--- RESPUESTA GROQ (Tardó: {tiempo_groq:.2f}s) ---")
    print(res_groq.choices[0].message.content) 

except Exception as e:
    print(f"Error en Groq: {e}")


# --- MEDICIÓN DE GEMINI (Ejercicio 4: Razonamiento complejo) --- 
start_gemini = time.time()
try:
    # Usamos gemini-2.5-flash para asegurar disponibilidad 
    res_gemini = client_gemini.models.generate_content(
        model="gemini-2.5-flash",
        contents=f"Da una explicación académica, detallada y con un ejemplo real sobre: {tema}. Responde en Markdown." 
    )
    end_gemini = time.time()
    tiempo_gemini = end_gemini - start_gemini
    
    print(f"\n--- RESPUESTA GEMINI (Tardó: {tiempo_gemini:.2f}s) ---")
    print(res_gemini.text) 
    
except Exception as e:
    print(f"Error en Gemini: {e}")
```

```python
import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()
client = Groq()

# 1. Iniciamos el historial con el rol 'system' (El Director) 
historial = [
    {
        "role": "system", 
        "content": "Eres un asistente técnico, sarcástico y breve. Recuerdas el nombre del usuario."
    }
]

print("--- CHAT INICIADO (Escribe 'salir' para terminar) ---")

while True:
    # 2. El usuario pregunta (El Cliente) 
    pregunta = input("\nTú: ")
    
    if pregunta.lower() == "salir":
        break

    # Añadimos la pregunta al historial
    historial.append({"role": "user", "content": pregunta})

    # 3. Enviamos TODO el historial para que tenga "memoria" 
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=historial
    )

    respuesta_ia = response.choices[0].message.content
    print(f"\nIA: {respuesta_ia}")

    # 4. GUARDAMOS la respuesta de la IA (El Assistant) 
    # Esto es vital para que en la próxima pregunta sepa qué respondió antes.
    historial.append({"role": "assistant", "content": respuesta_ia})
```
