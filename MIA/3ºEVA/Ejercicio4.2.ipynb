```python
import whisper
import os
# Elegimos el modelo (puedes cambiar "base" por "small" o "medium" según tu PC/GPU)
model = whisper.load_model("base")
print("Modelo cargado correctamente.")
```

```python
# Cambia "salida.mp3" por el nombre de TU archivo real
audio_path = "podcast.mp3" 

if not os.path.exists(audio_path):
    print(f"Error: El archivo {audio_path} no existe.")
else:
    # Ahora Whisper procesará tu video real 
    result = model.transcribe(audio_path, verbose=False)
    print("-" * 30)
    print("TRANSCRIPCIÓN FINALIZADA")
    print("-" * 30)
    print(result["text"]) # Aquí aparecerá lo que dice tu video 
```

```python
# Paso 5: Guardar la transcripción en un archivo de texto 
nombre_salida = "transcripcion_resultado.txt" 

# Abrimos el archivo en modo escritura ("w") con codificación utf-8 [cite: 39]
with open(nombre_salida, "w", encoding="utf-8") as f:
    f.write(result["text"]) 

print(f"La transcripción se ha guardado en: {nombre_salida}")
```
