```python

incomes = []
for index,row in marketing.iterrows():
    # si el valor de una fila income es string
    if (type(row['Income']) == str):
        # metemos en la lista ese valor de income quitando las comas y en formato float
        incomes.append(float(row['Income'].replace(",","")))
    else:
        # si no es string lo metemos igual
        incomes.append(row['Income'])

# igualamos las filas income a los valores de la lista
marketing['Income'] = incomes
```

Celda 10: Ejercicio 2.3 - Equipos con mal cálculo de puntos
- Muestra el nombre de los equipos que tienen mal calculados los puntos por partido. Es decir, que no coincida la columna de puntos por partido (PTS) con el cálculo: 3*3P + 2*2P + FT
- Crear una columna con los puntos calculados según la fórmula y mostrar los equipos donde no coincide.

```python
equipos = []
Puntos_Calculados = nba["3P"]*3 + nba["2P"]*2 + nba["FT"]

for index, row in nba.iterrows():
    calculado = row["3P"]*3 + row["2P"]*2 + row["FT"]
    if row["PTS"] != calculado:
        equipos.append(row["Team"])

nba["Puntos_Calculados"] = Puntos_Calculados

nba.head()
```
