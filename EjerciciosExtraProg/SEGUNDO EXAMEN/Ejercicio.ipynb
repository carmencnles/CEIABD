📉 Bloque de Regresión (Predecir Números)
1. Regresión Lineal (Simple y Múltiple)
Caso de uso: Predecir un valor numérico continuo (ej. precio de coches, índice de felicidad) cuando la relación es proporcional.
Parámetros / Hiperparámetros:
fit_intercept (bool): Si se calcula el punto donde la recta corta el eje Y (por defecto True).
Learning Rate (Teórico): En el descenso de gradiente, controla el tamaño del paso para encontrar el mínimo error.Métrica: $R^2$ (R-cuadrado).

2. Regresión Polinómica
Caso de uso: Cuando los datos hacen una curva (ej. casos de COVID, crecimiento no lineal).
Parámetros:
degree (int): El grado del polinomio. Ojo: grado alto = overfitting.
include_bias (bool): Si añade una columna de 1s (intercepto). 
Normalmente False si ya usas LinearRegression después.

Bloque de Clasificación (Predecir Categorías)
3. SVM (Support Vector Machines)
Caso de uso: Clasificar cuando las fronteras entre grupos están muy juntas. Ideal para datasets pequeños/medianos de flores o diagnóstico médico. Requiere escalado previo.

Parámetros:

C (float): El más importante. C bajo = margen ancho (permite fallos, mejor generalización). C alto = margen estrecho (intenta ser perfecto en el entrenamiento).

kernel: linear (rectas), poly (curvas), rbf (formas circulares/complejas).

Métrica: Accuracy, F1-Score y Matriz de Confusión.

4. Árboles de Decisión (Decision Tree)
Caso de uso: Clasificación basada en reglas lógicas (SI esto ES mayor que X, ENTONCES es clase A). Muy visual.

Parámetros:

max_depth (int): Límite de niveles. Fundamental para evitar que el árbol aprenda los datos de memoria.

criterion: gini (por defecto) o entropy. Mide la pureza de los nodos.

min_samples_leaf: Cuántos datos debe haber como mínimo al final de una rama.

5. Random Forest
Caso de uso: Cuando quieres la máxima precisión en clasificación (ej. reconocer dígitos escritos a mano como en MNIST). Es un conjunto de muchos árboles.

Parámetros:

n_estimators (int): Número de árboles en el bosque (ej. 100). A más árboles, más robusto pero más lento.

max_depth: Igual que en el árbol individual, para controlar el sobreajuste.
