```python
# Ejercicio 2 - 1

vinos = pd.read_csv("winequalityN.csv")

vinos.head()
```

```python
vinos.isnull().sum()
```

```python
# Para hacer desaparecer los nulos
# Voy a reemplazar los valores nulos por las medias de cada columna

media_fixed_acidity = vinos["fixed acidity"].mean()
media_volatile_acidity = vinos["volatile acidity"].mean()
media_citric_acid = vinos["citric acid"].mean()
media_residual_sugar = vinos["residual sugar"].mean()
media_chlorides = vinos["chlorides"].mean()
media_pH = vinos["pH"].mean()
media_sulphates = vinos["sulphates"].mean()
media_quality = vinos["quality"].mean()
```

```python
# 2.2 

nulos = []
for index, row in vinos.iterrows():
    if row["fixed acidity"] != row["fixed acidity"]:
        nulos.append(media_fixed_acidity)
    else:
        nulos.append(index)

vinos["fixed acidity"] = nulos

nulos1 = []
for index, row in vinos.iterrows():
    if row["volatile acidity"] != row["volatile acidity"]:
        nulos1.append(media_volatile_acidity)
    else:
        nulos1.append(index)

vinos["volatile acidity"] = nulos1

nulos2 = []
for index, row in vinos.iterrows():
    if row["citric acid"] != row["citric acid"]:
        nulos2.append(media_citric_acid)
    else:
        nulos2.append(index)

vinos["citric acid"] = nulos2


nulos3 = []
for index, row in vinos.iterrows():
    if row["residual sugar"] != row["residual sugar"]:
        nulos3.append(media_residual_sugar)
    else:
        nulos3.append(index)

vinos["residual sugar"] = nulos3


nulos4 = []
for index, row in vinos.iterrows():
    if row["chlorides"] != row["chlorides"]:
        nulos4.append(media_chlorides)
    else:
        nulos4.append(index)

vinos["chlorides"] = nulos4


nulos5 = []
for index, row in vinos.iterrows():
    if row["pH"] != row["pH"]:
        nulos5.append(media_pH)
    else:
        nulos5.append(index)

vinos["pH"] = nulos5


nulos6 = []
for index, row in vinos.iterrows():
    if row["sulphates"] != row["sulphates"]:
        nulos6.append(media_sulphates)
    else:
        nulos6.append(index)

vinos["sulphates"] = nulos6

nulos7 = []
for index, row in vinos.iterrows():
    if row["quality"] != row["quality"]:
        nulos7.append(media_quality)
    else:
        nulos7.append(index)

vinos["quality"] = nulos7


vinos.tail(20)
```
