Celda 4: Ejercicio 1.3 - Columna "Descuento_Aplicado"

- Recorre el DataFrame con un bucle for
- Crea una nueva columna llamada "Descuento_Aplicado"
- La columna será True cuando el precio de venta sea menor que el precio original
- La columna será False en caso contrario
- Recuerda hacerlo recorriendo el DataFrame
- Muestra cuántas ventas tuvieron descuento aplicado

```python
Descuento_Aplicado = []
for index, row in ventas.iterrows():
    if row["Precio_Venta"] < row["Precio_Original"]:
        Descuento_Aplicado.append(True)
    else:
        Descuento_Aplicado.append(False)

ventas["Descuento_Aplicado"] = Descuento_Aplicado

ventas.head()
```

```python
# SOLO ESTO (2 líneas):
conteo_true = ventas["Descuento_Aplicado"].sum()
print(f"Ventas con descuento aplicado: {conteo_true}")

# Si también quieres False:
conteo_false = (~ventas["Descuento_Aplicado"]).sum()
print(f"Ventas sin descuento: {conteo_false}")
```

```python
# Otra forma, mas enrevesada.

conteo = 0
for index, row in ventas.iterrows():
    if row['Descuento_Aplicado'] == True:
        conteo += 1
    else:
        None

print(f"Han tenido descuento {conteo} personas")
```

Celda 12: Ejercicio 3.1 - Gráfico de barras

- Agrupa las ventas por Plataforma y suma las Unidades_Vendidas
- Crea un gráfico de barras mostrando las ventas totales por cada Plataforma
- El gráfico debe tener título: "Ventas Totales por Plataforma"
- El eje X debe llamarse "Plataforma"
- El eje Y debe llamarse "Unidades Vendidas"
- Muestra el gráfico

```python
videojuegos = pd.read_csv("videojuegos.csv", sep=";")

videojuegos.head()
```

```python
unidades = videojuegos.groupby("Plataforma")["Unidades_Vendidas"].sum().to_frame()
```

```python
unidades.head()
```

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns


# Grafico de barras
plt.figure(figsize=(10, 6))
sns.barplot(x="Plataforma", y="Unidades_Vendidas", data=unidades, color="pink")
plt.xticks(rotation=45, ha='right')
plt.title("Venta totales por plataforma")
plt.xlabel("Plataforma")
plt.ylabel("Unidades Vendidas")
#plt.tight_layout()
plt.show()
```
