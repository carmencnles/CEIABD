```python
#1.1
import pandas as pd
marketing = pd.read_csv('marketing_campaign.csv',sep='\t')
marketing.head()
```

```python
#1.2 Cambia la columna dt_customer a formato Datetime y quédate con los 1000 clientes más modernos. 
# El cambio debe ser permanente y usarlo a partir de aquí en todo el examen.
import datetime
marketing['Dt_Customer'] = pd.to_datetime(marketing['Dt_Customer'],dayfirst=True)
marketing = marketing.sort_values(by='Dt_Customer',ascending=False).iloc[:1000]
```

```python
#1.3 Recorre el dataframe para generar una columna nueva llamada "Probably_Died". 
# El valor de está columna será True en los casos en los que el cliente haya nacido 
# hace más de 100 años. False en el resto de los casos.
probably_died = []
for index,row in marketing.iterrows():
    if row['Year_Birth'] < 1925:
        probably_died.append(True)
    else:
        probably_died.append(False)
marketing['Probably_Died'] = probably_died
marketing.sort_values(by='Year_Birth',ascending=True).head(20)
```

```python
marketing[['MntWines','MntFruits']]
```

```python
#1.4 Crea una función (llamada new_column) que reciba un dataframe y devuelva un pandas Series
#  con la suma de las columnas MntWines	MntFruits.
def new_column (df):
    pd_series = df['MntWines'] + df['MntFruits']
    return pd_series

print(new_column(marketing))
```

```python
#1.5 Convierte la columna Education para que sea numérica.
# Para ello debes usar df['Education'].unique() que te devolverá todos los valores únicos
# de la columna.
# Asigna un valor numérico a cada uno y sustituye en la columna todos por esos números. 
print(marketing['Education'].unique())
marketing.replace(to_replace='Graduation',value=0,inplace=True)
marketing.replace(to_replace='PhD',value=1,inplace=True)
marketing.replace(to_replace='Master',value=2,inplace=True)
marketing.replace(to_replace='2n Cycle',value=3,inplace=True)
marketing.replace(to_replace='Basic',value=4,inplace=True)
marketing.head(10)
```

```python
#1.6 Filtra el dataframe para guardar uno nuevo, llamado VIP_clients. En el tienen que figurar 
# solo los clientes que tengan MntWines mayor que 400 e income mayor que 80000. 
# Cuando te encuentres problemas, los tienes que solucionar usando el propio dataframe, no
# directamente en el csv. Dependiendo de como lo quieras hacer puede que tengas que comprobar 
# el tipo de algún objeto. Dicha comparación se hace así: type(objeto) == str

incomes = []
for index,row in marketing.iterrows():
    if (type(row['Income']) == str):
        incomes.append(float(row['Income'].replace(",","")))
    else:
        incomes.append(row['Income'])


marketing['Income'] = incomes
VIP_clients = marketing.loc[(marketing['MntWines']>400) & (marketing['Income']>80000)]
VIP_clients.head()
```

```python
# Muestra el nombre de los equipos que tienen mal calculados los puntos por partido. 
# Es decir, que no coincida la columna de puntos por partido (PTS) con el cálculo:
# 3*3P + 2*2P + FT
for index,row in nba.iterrows():
    if (float(row['PTS'])) != float(row['3P'])*3+float(row['2P'])*2+float(row['FT']):
        print(row['Team'])
```
