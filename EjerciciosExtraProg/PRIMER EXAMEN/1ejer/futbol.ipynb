Ejercicio 2.3: Crear columnas con condiciones

 1. Crea una columna 'Es_Veterano' que sea True si Edad > 30
 2. Crea una columna 'Categoria_Salario':
    - 'Alto' si salario > 30 millones
    - 'Medio' si salario entre 10 y 30 millones
    - 'Bajo' si salario < 10 millones
 3. Muestra solo: Jugador, Edad, Es_Veterano, Categoria_Salario

```python
Es_Veterano = []
for index,row in futbol.iterrows():
    if row['Edad'] > 30:
        Es_Veterano .append(True)
    else:
        Es_Veterano .append(False)
futbol['Es_Veterano'] = Es_Veterano
futbol.head(10)
```

```python
Categoria_Salario = []
for index,row in futbol.iterrows():
    if row['Salario'] > 30000000:
        Categoria_Salario .append("Alto")
    elif row['Salario'] > 10000000 and row['Salario'] < 30000000:
        Categoria_Salario .append("Medio")
    else:
        Categoria_Salario .append("Bajo")
futbol['Categoria_Salario'] = Categoria_Salario
futbol.head(100)
```
