```python
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
```

```python
portatiles = pd.read_csv('laptop_price.csv', encoding='latin-1')

portatiles.head()
```

```python
# Para sacar los valores de cada campo con unique()
valores_unicos = {col: portatiles[col].unique() for col in portatiles.columns}

for col, values in valores_unicos.items():
    print(f"Columna {col}: {values}")
```

```python
portatiles.info()
```

```python
portatiles.head()
```

```python
portatiles = portatiles.drop(columns=["laptop_ID", "Product"])
```

```python
portatiles["Weight"] = portatiles["Weight"].str.replace("kg", "").astype(float)
```

```python
portatiles = pd.get_dummies(portatiles,dtype='int')
```

```python
portatiles.corr()["Price_euros"].sort_values(ascending=False)
```

```python
portatiles.head()
```

```python
#scaler_X = StandardScaler()
#scaler_y = StandardScaler()
scaler = StandardScaler()

X = scaler.fit_transform(portatiles.drop(columns=["Price_euros"]))
y = scaler.fit_transform(portatiles[["Price_euros"]])

X_train_full, X_test, y_train_full, y_test = train_test_split(X, y)
X_train, X_valid, y_train, y_valid = train_test_split(X_train_full, y_train_full)
```

```python
model = keras.Sequential([
    keras.layers.Dense(180, activation="relu", input_shape=X_train.shape[1:]),
    keras.layers.Dense(80, activation="relu"),
    keras.layers.Dense(1)  
])

model.compile(optimizer=keras.optimizers.SGD(learning_rate=0.001), loss="mean_squared_error",metrics=['r2_score'])

history = model.fit(X_train, y_train, epochs=100,
validation_data=(X_valid, y_valid))
```

```python
y_pred = model.predict(X_test)
```

```python
r2 = r2_score(y_test, y_pred)

print(r2)
```
