```python
import pandas as pd
pegatinas = pd.read_csv('stickers.csv')
pegatinas.head()
```

```python
#Separa los años y meses en columnas categóricas. 
#Elimina las columnas que debes eliminar
#Prepara los datos para poder lanzar los algoritmos de machine learning.
#Las 10 variables con más correlación con num_sold.
#Prueba con SVM y con DecissionTree.

#1
years = []
months = []
for i,row in pegatinas.iterrows():
    years.append(row['date'].split('-')[0])
    months.append(row['date'].split('-')[1])
pegatinas['year'] = years
pegatinas['month'] = months

pegatinas.head()
```

```python
pegatinas.shape
```

```python
pegatinas.drop(['id','date','city'],axis=1,inplace=True)
pegatinas.head()
```

```python
pegatinas.head()
```

```python
pegatinas.dropna(inplace=True)
```

```python
pegatinas.info()
```

```python
pegatinas = pd.get_dummies(pegatinas,columns=['country','store','product','year','month'],dtype='int')
pegatinas.head()
```

```python
pegatinas.columns
```

```python
pegatinas.corr()['num_sold'].abs().sort_values(ascending=False)[1:10]
```

```python
X = pegatinas.drop(['num_sold'],axis=1)
y = pegatinas['num_sold'].to_frame()
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = scaler.fit_transform(X)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
#APARTADO 2
from sklearn.svm import SVR
svr_rbf = SVR(kernel="rbf", C=60)
#svr_lnr = SVR(kernel="linear", C=60)
#svr_poly = SVR(kernel="poly", degree=1, C=60)
svr_rbf.fit(X_train,y_train)
y_pred_rbf = svr_rbf.predict(X_test)
#svr_lnr.fit(X_train,y_train)
#y_pred_lnr = svr_lnr.predict(X_test)
#svr_poly.fit(X_train,y_train)
#y_pred_poly = svr_poly.predict(X_test)
from sklearn.metrics import r2_score
print(r2_score(y_test, y_pred_rbf))
#print(r2_score(y_test, y_pred_lnr))
#print(r2_score(y_test, y_pred_poly))
```

```python
#APARTADO 3
from sklearn.tree import DecisionTreeRegressor
#tree_reg = DecisionTreeRegressor(random_state=0,max_depth=9,min_samples_split=15)
tree_reg = DecisionTreeRegressor(random_state=0)
tree_reg.fit(X_train, y_train)
y_pred_tree = tree_reg.predict(X_test)
from sklearn.metrics import r2_score
print(r2_score(y_test, y_pred_tree))
```
