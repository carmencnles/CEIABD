```python
svm_rbf = SVC(kernel="rbf", C=1.0, gamma=3.0, random_state=42)
svm_rbf.fit(X_train_scaled, y_train)
```

```python
pred_lineal = svm_lineal.predict(X_test_scaled)
pred_poly = svm_poly.predict(X_test_scaled)
pred_rbf = svm_rbf.predict(X_test_scaled)


print("=== SVM LINEAL ===")
print("Accuracy:", accuracy_score(y_test, pred_lineal)) #exactitud
#print(confusion_matrix(y_test, pred_lineal))
#print(classification_report(y_test, pred_lineal))

print("\n=== SVM POLY ===")
print("Accuracy:", accuracy_score(y_test, pred_poly))
#print(confusion_matrix(y_test, pred_poly))
#print(classification_report(y_test, pred_poly))

print("\n=== SVM RBF ===")
print("Accuracy:", accuracy_score(y_test, pred_rbf))
#print(confusion_matrix(y_test, pred_rbf))
#print(classification_report(y_test, pred_rbf))
```

```python
estudiantes["Target"] = estudiantes["Target"].replace({
    "Dropout": 2,
    "Graduate": 1,
    "Enrolled": 0
}).astype(int)
```

```python
estudiantes.info()
```

```python
estudiantes.corr()["Target"].abs().sort_values(ascending=False)[1:]
```

```python
X = estudiantes.drop(columns=["Target"])
y = estudiantes['Target']
```

```python
# Separo X train de X test

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

X_train.shape, X_test.shape
```

```python
# Escalo los datos con StandardScaler.

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
svm_lineal = SVC(kernel="linear", C=1.0, random_state=42)
svm_lineal.fit(X_train_scaled, y_train)
```

```python
svm_poly = SVC(kernel="poly", C=1.0, gamma=1.0, degree=3, random_state=42)
svm_poly.fit(X_train_scaled, y_train)
```

```python
svm_rbf = SVC(kernel="rbf", C=1.0, gamma=3.0, random_state=42)
svm_rbf.fit(X_train_scaled, y_train)
```

```python
pred_lineal = svm_lineal.predict(X_test_scaled)
pred_poly = svm_poly.predict(X_test_scaled)
pred_rbf = svm_rbf.predict(X_test_scaled)


print("=== SVM LINEAL ===")
print("Accuracy:", accuracy_score(y_test, pred_lineal)) #exactitud
#print(confusion_matrix(y_test, pred_lineal))
#print(classification_report(y_test, pred_lineal))

print("\n=== SVM POLY ===")
print("Accuracy:", accuracy_score(y_test, pred_poly))
#print(confusion_matrix(y_test, pred_poly))
#print(classification_report(y_test, pred_poly))

print("\n=== SVM RBF ===")
print("Accuracy:", accuracy_score(y_test, pred_rbf))
#print(confusion_matrix(y_test, pred_rbf))
#print(classification_report(y_test, pred_rbf))
```

```python
tree_reg = DecisionTreeRegressor(max_depth=15, min_samples_split=200, min_samples_leaf=100,random_state=42)
tree_reg.fit(X_train, y_train)
```

```python
y_pred = tree_reg.predict(X_test)
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```
