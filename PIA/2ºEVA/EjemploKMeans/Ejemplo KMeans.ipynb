## Ejemplo KMeans
<div style="border-style:groove;border-width:thin;padding:10px">
Vamos a ver el primer ejemplo de esta técnica de clustering. Para ello vamos a usar el dataset de las flores pero sin las etiquetas del tipo de flor. Recordamos que este algoritmo agrupa los datos por similitudes teniendo en cuenta el número de clusters o grupos que le pasamos nosotros inicialmente. 
    </div>

<div style="border-style:groove;border-width:thin;padding:10px">
Lo primero que vamos a hacer es importar los datos del dataset de las flores (iris). Recordamos un poco como era.
    </div>

```python
import numpy as np
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report

iris = datasets.load_iris()
print(type(iris))
print(iris)
print(iris.DESCR)
```

```python
flores = pd.DataFrame()
flores['tipo_flor'] = iris['target']
flores['sepal_length'] = iris['data'][:,0]
flores['sepal_width'] = iris['data'][:,1]
flores['petal_length'] = iris['data'][:,2]
flores['petal_width'] = iris['data'][:,3]
print(flores.head())

flores.corr()
```

```python
plt.scatter(flores.loc[flores['tipo_flor']==0]['petal_length'],flores.loc[flores['tipo_flor']==0]['petal_width'],color='y')
plt.scatter(flores.loc[flores['tipo_flor']==1]['petal_length'],flores.loc[flores['tipo_flor']==1]['petal_width'],color='r')
plt.scatter(flores.loc[flores['tipo_flor']==2]['petal_length'],flores.loc[flores['tipo_flor']==2]['petal_width'],color='b')

plt.show()
```

```python
X = flores[['petal_length','petal_width']]
#y = flores['tipo_flor'].to_list()
```

<div style="border-style:groove;border-width:thin;padding:10px">
1. Importa el KMeans del paquete cluster de sklearn y crea el objeto usando el parámetro n_clusters=3. Por suerte en este ejercicio sabemos que hay 3 tipos de flores. Ahora entrena y etiqueta cada dato de entrada de X con el método fit_predict (guarda el resultado de este método en una variable llamada y_pred).
</div>

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Creamos el escalador
scaler = StandardScaler()

# 2. Ajustamos y transformamos tus datos X
# Esto convierte tus valores originales en puntuaciones Z (Z-scores)
X_scaled = scaler.fit_transform(X)

# 3. Ahora usamos X_scaled en el KMeans
kmeans_scaled = KMeans(n_clusters=3, random_state=42, n_init="auto")
y_pred_scaled = kmeans_scaled.fit_predict(X_scaled)

# Los centroides ahora también estarán en unidades "escaladas"
centroids_scaled = kmeans_scaled.cluster_centers_
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

# Usamos X_scaled para los ejes (0 es petal_length, 1 es petal_width)
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=y_pred_scaled, cmap='viridis', alpha=0.6)

# Pintamos los centroides escalados
plt.scatter(centroids_scaled[:, 0], centroids_scaled[:, 1], 
            marker='X', s=200, color='red', label='Centros Escalados')

plt.title('K-Means con Datos Escalados')
plt.xlabel('Petal Length (Estandarizado)')
plt.ylabel('Petal Width (Estandarizado)')
plt.show()
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3, random_state=42, n_init="auto")
kmeans.fit(X)
```

```python
y_pred = kmeans.fit_predict(X)
```

<div style="border-style:groove;border-width:thin;padding:10px">
2. Mete el atributo cluster_centers_ en una variable y pinta con scatter los centros para ver que tal han quedado ubicados.
</div>

```python
centroids = kmeans.cluster_centers_
plt.scatter(flores.loc[flores['tipo_flor']==0]['petal_length'],flores.loc[flores['tipo_flor']==0]['petal_width'],color='y')
plt.scatter(flores.loc[flores['tipo_flor']==1]['petal_length'],flores.loc[flores['tipo_flor']==1]['petal_width'],color='r')
plt.scatter(flores.loc[flores['tipo_flor']==2]['petal_length'],flores.loc[flores['tipo_flor']==2]['petal_width'],color='b')

plt.scatter(centroids[:, 0], centroids[:, 1], marker='X', s=200, color='black', label='Centroids')

plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.legend()
plt.show()

print(f"Centroids shape: {centroids.shape}")
```

```python
import matplotlib.pyplot as plt

# Extraemos los centroides de nuevo por si acaso
centroids = kmeans.cluster_centers_

plt.figure(figsize=(8, 6))

# Pintamos todos los puntos a la vez
# Usamos el parámetro 'c' para que los colores dependan de la predicción
plt.scatter(flores['petal_length'], flores['petal_width'], c=y_pred, cmap='viridis', s=50, alpha=0.7)

# Pintamos los centroides (usando índices 0 y 1 basados en tu error anterior)
plt.scatter(centroids[:, 0], centroids[:, 1], marker='X', s=250, color='red', label='Centroides')

plt.title('Agrupación K-Means (Predicción)')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.colorbar(label='Clúster asignado')
plt.legend()
plt.show()
```

    </div>

<div style="border-style:groove;border-width:thin;padding:10px">
4. Ahora cambia los parámetros de KMeans. Pon init='random'm dale n_clusters=4, n_init=1 para que lo haga una sola vez y vete cambiando el random_state. Pinta los centros como hicimos antes.
    </div>

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

mi_random_state = 20  

kmeans_diferente = KMeans(n_clusters=4, 
                     init='random', 
                     n_init=1, 
                     random_state=mi_random_state)

# 2. Entrenamos y obtenemos predicciones
y_pred_loco = kmeans_diferente.fit_predict(X)
centros_locos = kmeans_diferente.cluster_centers_

# 3. Pintamos los resultados
plt.figure(figsize=(8, 6))

# Pintamos los puntos coloreados por la nueva predicción (ahora hay 4 colores)
plt.scatter(flores['petal_length'], flores['petal_width'], c=y_pred_loco, cmap='plasma', s=50, alpha=0.6)

# Pintamos los 4 centros (usando índices 0 y 1 como antes)
plt.scatter(centros_locos[:, 0], centros_locos[:, 1], marker='X', s=300, color='cyan', edgecolors='black', label='Centros Aleatorios')

plt.title(f'K-Means con n_clusters=4 y random_state={mi_random_state}')
plt.legend()
plt.show()
```
