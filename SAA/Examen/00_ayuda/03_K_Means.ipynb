```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# -----------------------------
# 1. Crear dataset simulado
# -----------------------------
np.random.seed(42)

df = pd.DataFrame({
    "ingresos": np.random.randint(1000, 5000, 200),
    "gasto": np.random.randint(100, 1000, 200)
})

# -----------------------------
# 2. Seleccionar variables
# -----------------------------
X = df[["ingresos", "gasto"]]

# -----------------------------
# 3. Aplicar K-Means
# -----------------------------
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

labels = kmeans.labels_
centroides = kmeans.cluster_centers_

# -----------------------------
# 4. Mostrar centroides
# -----------------------------
print(" Coordenadas de los centroides:\n")
for i, c in enumerate(centroides):
    print(f"Centroide {i}: ingresos={c[0]:.2f}, gasto={c[1]:.2f}")

# -----------------------------
# 5. Visualización
# -----------------------------
plt.figure(figsize=(7,6))

# puntos
plt.scatter(X["ingresos"], X["gasto"], c=labels, s=30, alpha=0.6)

# centroides
plt.scatter(centroides[:,0], centroides[:,1],
            c='black', marker='X', s=150, label="Centroides")

# etiquetas
for i, c in enumerate(centroides):
    plt.text(c[0], c[1], f"C{i}", fontsize=12, ha='center')

plt.xlabel("Ingresos")
plt.ylabel("Gasto")
plt.title("Segmentación de clientes con K-Means")
plt.legend()
plt.show()
```
