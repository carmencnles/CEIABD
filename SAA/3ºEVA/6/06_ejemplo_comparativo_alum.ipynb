# COMPARATIVA APRENDIZAJE SUPERVISADO - APRENDIZAJE NO SUPERVISADO

```python
# Importaciones
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score
```

## Dataset

```python
X, y = make_blobs(n_samples=300, centers=3, random_state=42)

plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Dataset con etiquetas reales")
plt.show()
```

## Modelo Supervisado

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

modelo_sup = LogisticRegression()
modelo_sup.fit(X_train, y_train)

y_pred = modelo_sup.predict(X_test)

print("Accuracy supervisado:", accuracy_score(y_test, y_pred))
```

## Visualización Modelo Supervisado

```python
plt.scatter(X_test[:,0], X_test[:,1], c=y_pred)
plt.title("Predicciones modelo supervisado")
plt.show()
```

## Modelo k - Means

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

clusters = kmeans.labels_
```

## Visualización K - Means

```python
plt.scatter(X[:,0], X[:,1], c=clusters)
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], 
            color='red', marker='X', s=200)

plt.title("Clusters encontrados por K-Means")
plt.show()
```

## Comparación

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Etiquetas reales")

plt.subplot(1,2,2)
plt.scatter(X[:,0], X[:,1], c=clusters)
plt.title("Clusters K-Means")

plt.show()
```
