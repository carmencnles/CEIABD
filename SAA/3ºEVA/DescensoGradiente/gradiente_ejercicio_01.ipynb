# EJEMPLO GRADIENTE

## 0. Librerías

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. ALgoritmo

>El descenso del gradiente busca el mínimo de una función moviéndose en la dirección opuesta al gradiente.

```python
def gradient_descent(f, grad, x0, alpha=0.1, iteraciones=20):
    
    x = x0
    xs = [x]

    for i in range(iteraciones):
        x = x - alpha * grad(x)
        xs.append(x)

    return xs
```

### 1.1 Definiciones

```python
# ---- FUNCIONES A OPTIMIZAR ----

# 1. Función cuadrática
def f_cuadratica(x):
    return x**2

def grad_cuadratica(x):
    return 2*x


# 2. Función con mínimo plano
def f_polinomio4(x):
    return x**4

def grad_polinomio4(x):
    return 4*x**3


# 3. Función con mínimos locales
def f_minimos_locales(x):
    return x**4 - 3*x**3 + 2

def grad_minimos_locales(x):
    return 4*x**3 - 9*x**2
```

## 2. Función cuadrática (Polinomio de grado dos)

```python


xs = gradient_descent(f_cuadratica, grad_cuadratica, x0=4, alpha=0.1)

x_vals = np.linspace(-5,5,200)

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_cuadratica(x_vals), label="f(x)=x²")
plt.scatter(xs, f_cuadratica(np.array(xs)), color="red")
plt.plot(xs, f_cuadratica(np.array(xs)), linestyle="--")

plt.title("Descenso del gradiente - función cuadrática")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.ylim(-1, 17)
plt.xlim(-5, 5)
plt.show()
```

## 3. Polinomio de grado 4

```python


xs = gradient_descent(f_polinomio4, grad_polinomio4, x0=2, alpha=0.05)

x_vals = np.linspace(-3,3,200)

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_polinomio4(x_vals), label="f(x)=x⁴")
plt.scatter(xs, f_polinomio4(np.array(xs)), color="red")
plt.plot(xs, f_polinomio4(np.array(xs)), linestyle="--")

plt.title("Descenso del gradiente - convergencia lenta")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.ylim(-1, 20)
plt.xlim(-3, 3)
plt.show()
```

## 4. Función con mínimos locales

```python


xs = gradient_descent(f_minimos_locales, grad_minimos_locales, x0=0.5, alpha=0.01)

x_vals = np.linspace(-1,3,200)

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_minimos_locales(x_vals), label="f(x)=x⁴-3x³+2")
plt.scatter(xs, f_minimos_locales(np.array(xs)), color="red")
plt.plot(xs, f_minimos_locales(np.array(xs)), linestyle="--")

plt.title("Descenso del gradiente - mínimos locales")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.ylim(-10, 10)
plt.xlim(-1, 3.5)
plt.show()
```

## 5. Comparamos los tres casos

```python
fig, axs = plt.subplots(1,3, figsize=(15,4))

# ---- CASO 1 ----
def f(x):
    return x**2

def grad(x):
    return 2*x

xs = gradient_descent(f, grad, x0=4, alpha=0.1)

x_vals = np.linspace(-5,5,200)

axs[0].plot(x_vals, f(x_vals))
axs[0].scatter(xs, f(np.array(xs)), color="red")
axs[0].plot(xs, f(np.array(xs)), linestyle="--")
axs[0].set_title("f(x)=x²")
axs[0].grid()


# ---- CASO 2 ----
def f(x):
    return x**4

def grad(x):
    return 4*x**3

xs = gradient_descent(f, grad, x0=2, alpha=0.05)

x_vals = np.linspace(-3,3,200)

axs[1].plot(x_vals, f(x_vals))
axs[1].scatter(xs, f(np.array(xs)), color="red")
axs[1].plot(xs, f(np.array(xs)), linestyle="--")
axs[1].set_title("f(x)=x⁴ (mínimo plano)")
axs[1].grid()


# ---- CASO 3 ----
def f(x):
    return x**4 - 3*x**3 + 2

def grad(x):
    return 4*x**3 - 9*x**2

xs = gradient_descent(f, grad, x0=0.5, alpha=0.01)

x_vals = np.linspace(-1,3,200)

axs[2].plot(x_vals, f(x_vals))
axs[2].scatter(xs, f(np.array(xs)), color="red")
axs[2].plot(xs, f(np.array(xs)), linestyle="--")
axs[2].set_title("mínimos locales")
axs[2].grid()


plt.suptitle("Comparación del descenso del gradiente en distintas funciones")

plt.show()
```
