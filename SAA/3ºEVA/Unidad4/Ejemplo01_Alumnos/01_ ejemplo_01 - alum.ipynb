# Ejemplo
Una tienda quiere agrupar a sus clientes según su comportamiento de compra, pero no tiene etiquetas previas.
Solo dispone de dos variables:

Gasto anual (€)

Frecuencia de compra (visitas al mes)

Se quiere descubrir si existen tipos de clientes.

```python
import numpy as np

X = np.array([
    [200, 2],
    [220, 3],
    [250, 2],
    [800, 8],
    [820, 9],
    [780, 7],
    [400, 4],
    [420, 5],
    [390, 4]
])
```

## ANÁLISIS DE LOS DATOS


1. ¿Este problema es supervisado o no supervisado? ¿Por qué?

2. ¿Qué crees que significa “agrupar clientes” en este contexto?

3. Observando los datos, ¿cuántos grupos crees que hay aproximadamente?

## VISUALIZACIÓN DE LOS DATOS

```python
import matplotlib.pyplot as plt

plt.scatter(X[:,0], X[:,1])
plt.xlabel("Gasto anual")
plt.ylabel("Frecuencia")
plt.title("Clientes")
plt.show()
```

## APLICACIÓN DE K - MEANS

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

labels = kmeans.labels_
centroides = kmeans.cluster_centers_
# para evitar problemas warnings.filterwarnings("ignore") justo depués de los primeros imports
```

## CLÚSTERS

```python
plt.scatter(X[:,0], X[:,1], c=labels)
plt.scatter(centroides[:,0], centroides[:,1], marker='X', s=200)
plt.title("Clusters detectados")
plt.show()
```

## IDENTIFICACIÓN

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

labels = kmeans.labels_
centroides = kmeans.cluster_centers_
```

```python
print(centroides)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X[:,0], X[:,1], c=labels, s=10)
plt.scatter(centroides[:,0], centroides[:,1],
            c='black', marker='X', s=100)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Mostrar coordenadas de los centroides
print(" Coordenadas de los centroides:\n")
for i, c in enumerate(centroides):
    print(f"Centroide {i}: {c}")

# Gráfico
plt.figure(figsize=(6,5))

# Puntos coloreados por cluster
plt.scatter(X[:,0], X[:,1], c=labels, s=10)

# Centroides
plt.scatter(centroides[:,0], centroides[:,1],
            c='black', marker='X', s=120, label="Centroides")

# Etiquetas de centroides
for i, c in enumerate(centroides):
    plt.text(c[0], c[1], f"C{i}", fontsize=12, ha='center', va='center')

plt.title("K-Means: clusters y centroides")
plt.legend()
plt.show()
```
