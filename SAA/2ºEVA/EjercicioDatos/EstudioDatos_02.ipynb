# Análisis de Datos con el Dataset "Adult Income"

El Dataset "Adult Income" es una colección de datos demográficos y laborales extraídos del censo estadounidense de 1994, recopilado por la Oficina del Censo de los Estados Unidos. Este conjunto de datos es ampliamente utilizado en tareas de aprendizaje automático, estadística y análisis de datos para estudios relacionados con clasificación, regresión y pruebas de hipótesis estadísticas.

El objetivo principal asociado con este dataset es predecir si un individuo gana más de $50,000 al año en función de sus atributos personales y laborales. Este análisis permite explorar relaciones entre variables y realizar interpretaciones significativas a nivel social y económico.

## Descripción de las Variables

El dataset incluye las siguientes variables:

- **age**: Edad del individuo (en años).
- **workclass**: Tipo de empleo (p. ej., privado, gubernamental, autónomo, etc.).
- **fnlwgt**: Peso final asignado a la observación, utilizado para expandir la muestra del censo.
- **education**: Nivel educativo alcanzado (p. ej., secundaria, licenciatura, maestría).
- **education.num**: Nivel educativo en formato numérico (años de educación).
- **marital.status**: Estado civil (p. ej., casado, soltero, divorciado).
- **occupation**: Ocupación laboral (p. ej., administrativo, técnico, etc.).
- **relationship**: Relación familiar con el cabeza de familia (p. ej., esposo/a, hijo/a, etc.).
- **race**: Raza del individuo (p. ej., blanca, negra, asiática).
- **sex**: Género del individuo (masculino o femenino).
- **capital.gain**: Ganancia de capital obtenida.
- **capital.loss**: Pérdida de capital reportada.
- **native.country**: País de origen.
- **hours.per.week**: Número de horas trabajadas por semana.
- **income**: Variable objetivo, indica si el ingreso anual es '>50K' o '<=50K'.

```python
# Selección de variables.

adultos_organizados = adultos[["age", "workclass", "education", "occupation", "sex", "capital.gain", "capital.loss", "hours.per.week", "income"]]

adultos_organizados.head()
```

```python
adultos_organizados_limpio = adultos_organizados[["age", "workclass", "education", "occupation", "sex", "capital.gain", "capital.loss", "hours.per.week", "income"]]

# adultos_organizados_limpio.head()
```

```python
# Limpieza de valores 0 en capital.gain

indices_a_borrar = adultos_organizados_limpio[adultos_organizados_limpio["capital.gain"] == 0].index
adultos_organizados_limpio.drop(indices_a_borrar, inplace=True)

adultos_organizados_limpio.head()
```

```python
interrogaciones_a_borrar = adultos_organizados_limpio[(adultos_organizados_limpio["workclass"] == "?") & (adultos_organizados_limpio['occupation']== "?")].index
adultos_organizados_limpio.drop(interrogaciones_a_borrar, inplace=True)
adultos_organizados_limpio.head(100)
```

```python
# Detectar outliers

import numpy as np
import matplotlib.pyplot as plt

# Limite inferior = q1 - (1.5 * iqr)
# Limite superior = q3 + (1.5 * iqr)

q1 = np.quantile(adultos_organizados_limpio["capital.gain"], 0.25)
q2 = np.quantile(adultos_organizados_limpio["capital.gain"], 0.50)
q3 = np.quantile(adultos_organizados_limpio["capital.gain"], 0.75)
iqr = q3 - q1

limite_inferior = q1 - (1.5 * iqr)
limite_superior = q3 + (1.5 * iqr)
outliers = adultos_organizados_limpio["capital.gain"][(adultos_organizados_limpio["capital.gain"] < limite_inferior) | (adultos_organizados_limpio["capital.gain"]> limite_superior)]

print(f"Q1: {q1}")
print(f"Q2 (Mediana): {q2}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
print(f"Outliers detectados: {outliers}")
print(f"Limite inferior: {limite_inferior}")
print(f"Limite superior: {limite_superior}")


plt.xscale('log')
plt.boxplot(adultos_organizados_limpio["capital.gain"], vert=False)
plt.title("Boxplot de Visualizaciones")
plt.xlabel("Número de visualizaciones")
plt.grid(True)
plt.show()
```

```python
#Calcular medidas de tendencia central (media, mediana, moda) y dispersión (desviación estándar, varianza)
import statistics as stat

print("Age:")
print(np.median(adultos_organizados["age"]))
print(np.mean(adultos_organizados["age"]))
print(stat.mode(adultos_organizados["age"]))

print("Capital Gain:")
print(np.median(adultos_organizados["capital.gain"]))
print(np.mean(adultos_organizados["capital.gain"]))
print(stat.mode(adultos_organizados["capital.gain"]))

print("Capital Loss:")
print(np.median(adultos_organizados["capital.loss"]))
print(np.mean(adultos_organizados["capital.loss"]))
print(stat.mode(adultos_organizados["capital.loss"]))

print("Hours per week:")
print(np.median(adultos_organizados["hours.per.week"]))
print(np.mean(adultos_organizados["hours.per.week"]))
print(stat.mode(adultos_organizados["hours.per.week"]))

print(np.var(adultos_organizados["age"])) # Varianza
print(np.std(adultos_organizados["age"])) # Desviación estándar

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.pairplot(adultos_organizados, x_vars=['capital.gain', 'capital.loss', 'hours.per.week'], y_vars='age', height=4, aspect=1, kind='reg')
plt.show()
```

```python
plt.hist(adultos_organizados["sex"], bins=5, color='blue', edgecolor='black')
plt.title("Histograma de genero")
plt.xlabel("Edad")
plt.ylabel("Cantidad")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x="age", y="capital.gain", data=adultos_organizados)
plt.title("Relación entre Edad y Capital ganado")
plt.xlabel("Edad")
plt.ylabel("Capital")
plt.show()
```

```python
gana_mas_de_50K = []
for index,row in adultos_organizados.iterrows():
    if row['income'] == ">50K":
        gana_mas_de_50K .append(True)
    else:
        gana_mas_de_50K .append(False)
adultos_organizados['Gana_mas_de_50K'] = gana_mas_de_50K 
adultos_organizados.head(10)
```

```python
sns.boxplot(x='sex', y='Gana_mas_de_50K', data=adultos_organizados)
plt.title('Diagrama de Dispersión')
plt.show()
```

```python
conteo = adultos_organizados.groupby(['sex', 'Gana_mas_de_50K']).size().unstack(fill_value=0)

conteo.plot(kind='bar', figsize=(10, 6))
plt.title('Distribución de Ingresos por Sexo')
plt.xlabel('Sexo')
plt.ylabel('Cantidad de Personas')
plt.legend(['Gana ≤50K', 'Gana >50K'], title='Ingresos')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Análisis estadístico descriptivo
Calcula, para las variables numéricas seleccionadas:
- Medidas de tendencia central (media, mediana y moda).
- Medidas de dispersión (varianza y desviación estándar).
- Interpreta los resultados obtenidos, comentando qué indican sobre la distribución y variabilidad de los datos.

```python
# Medidas de tendencia central
import numpy as np
import statistics as stat

print("Medidas de tendencia central: Media, mediana y moda de age: ")
print(np.mean(adultos_organizados["age"]))
print(np.mean(adultos_organizados["age"]))
print(stat.mode(adultos_organizados["age"]))


print("Medidas de dispersión: varianza y desviación estándar de age: ")
varianza_muestral = np.var(adultos_organizados["age"], ddof=1)
print(f"Varianza: {varianza_muestral}")
desviacion_std = np.std(adultos_organizados["age"])
print(f"Desviación Estándar: {desviacion_std}")


print("Medidas de tendencia central: Media, mediana y moda de capital.gain: ")
print(np.mean(adultos_organizados["capital.gain"]))
print(np.mean(adultos_organizados["capital.gain"]))
print(stat.mode(adultos_organizados["capital.gain"]))


print("Medidas de dispersión: varianza y desviación estándar de capital.gain: ")
varianza_muestral = np.var(adultos_organizados["capital.gain"], ddof=1)
print(f"Varianza: {varianza_muestral}")
desviacion_std = np.std(adultos_organizados["capital.gain"])
print(f"Desviación Estándar: {desviacion_std}")



print("Medidas de tendencia central: Media, mediana y moda de hours.per.week: ")
print(np.mean(adultos_organizados["hours.per.week"]))
print(np.mean(adultos_organizados["hours.per.week"]))
print(stat.mode(adultos_organizados["hours.per.week"]))


print("Medidas de dispersión: varianza y desviación estándar de hours.per.week: ")
varianza_muestral = np.var(adultos_organizados["hours.per.week"], ddof=1)
print(f"Varianza: {varianza_muestral}")
desviacion_std = np.std(adultos_organizados["hours.per.week"])
print(f"Desviación Estándar: {desviacion_std}")

```
