3. Cuartiles e IQR
Con visualizaciones = [120, 130, 150, 160, 200, 210, 300, 320]:
1. Calcula Q1, Q2, Q3.
2. Calcula el IQR.
3. Detecta outliers.
4. Representa un boxplot

```python
import numpy as np
import matplotlib.pyplot as plt
visualizaciones_lista = [120, 130, 150, 160, 200, 210, 300, 320, 410, 4000]
visualizaciones = np.array(visualizaciones_lista) 
# Limite inferior = q1 - (1.5 * iqr)
# Limite superior = q3 + (1.5 * iqr)

q1 = np.quantile(visualizaciones, 0.25)
q2 = np.quantile(visualizaciones, 0.50)
q3 = np.quantile(visualizaciones, 0.75)
iqr = q3 - q1

limite_inferior = q1 - (1.5 * iqr)
limite_superior = q3 + (1.5 * iqr)
outliers = visualizaciones[(visualizaciones < limite_inferior) | (visualizaciones > limite_superior)]

print(f"Q1: {q1}")
print(f"Q2 (Mediana): {q2}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
print(f"Outliers detectados: {outliers}")
print(f"Limite inferior: {limite_inferior}")
print(f"Limite superior: {limite_superior}")


plt.xscale('log')
plt.boxplot(visualizaciones, vert=False)
plt.title("Boxplot de Visualizaciones")
plt.xlabel("Número de visualizaciones")
plt.grid(True)
plt.show()



```

5. Histograma y asimetría
Con tiempos = [100, 102, 105, 110, 115, 120, 200, 220, 250, 300]:
1. Histograma.
2. Asimetría.
3. Interpretación.

```python
import seaborn as sns

from scipy.stats import skew

tiempos = [100, 102, 105, 110, 115, 120, 200, 220, 250, 300]

plt.hist(tiempos, bins=5, color='blue', edgecolor='black')
plt.title("Histograma de Tiempos")
plt.xlabel("Tiempo")
plt.ylabel("Frecuencia")
plt.show()

#Asimetria 

print(f"Asimetría: {skew(tiempos)}") 


# Como el valor es positivo (0.75 > 0), tenemos una Asimetría Positiva (o a la derecha).
# Esto indica que la "cola" de la distribución se extiende hacia los valores más altos (200, 250, 300). 
# La mayoría de los datos están concentrados en la parte baja (entre 100 y 120).
# La Media (172.2) es considerablemente mayor que la Mediana (117.5). 
# Esto confirma que los valores altos están "arrastrando" el promedio hacia arriba, haciendo que no sea la mejor medida para representar al grupo típico.
```

6. Estudio de Outliers

3. Encuentra y justifica un valor adecuado para sustituir los outliers.
